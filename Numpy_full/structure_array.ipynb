```python
import numpy as np

people = np.array([
    ('Aman', 20, 82.5),
    ('Riyya', 19, 91.0),
    ('karan', 21, 79.5)
], dtype=[('name', 'U10'), ('age', 'i4'), ('score', 'f4')])

print(people['name'])

print(people['age'])

print(people[people['score'] > 80])
```
