# SORTING SEARCHING AND UNIQUE VALLUES

```python
import numpy as np
```

```python
a = np.array([30,10,40,20,30,10,20])
```

```python
print(np.sort(a))
```

```python
print(a)
```

```python
a.sort()
```

```python
print(a)
```

```python
print(np.argsort([30,10,40]))
```

```python
print(np.argmax(a))
```

```python
print(np.argmin(a))
```

```python
print(np.where(a == 20))
```

```python
print(np.unique(a))
```

```python
print(np.searchsorted(np.array([10,20,30,40]),25))
```

```python
print(np.nonzero(a))
```

```python
print(np.count_nonzero(a == 10))
```
