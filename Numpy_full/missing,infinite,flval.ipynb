# MISSING , INFINITE AND FOATING POINT VALUES

```python
import numpy as np
```

```python
a = np.array([1.0, np.nan, np.inf, -np.inf, 5.0])
```

```python
print(np.isnan(a))
```

```python
print(np.isinf(a))
```

```python
print(np.isfinite(a))
```

```python
print(np.nanmean(a)) #mean ignoring nan
```

```python
print(np.nan_to_num(a)) #replace nan/ infinite with finite value
```

```python
print(np.isclose(0.1 + 0.2, 0.3)) #tolerant float comparison
```
