JOINING AND SPILLITING


```python
import numpy as np
```

```python
a = np.array([1, 2])
```

```python
b = np.array([3,4])
```

```python
print(np.concatenate((a, b)))
```

```python
print(np.stack((a, b)))
```

```python
print(np.vstack((a, b)))
```

```python
print(np.hstack((a,b)))
```

```python
x = np.array([[1,2], [3,4]])
y = np.array([[5,6],[7,8]])
```

```python
print(np.concatenate((x, y), axis = 0)) # apend rows
```

```python
print(np.concatenate((x, y), axis=1)) # append column
```

```python
print(np.split(np.arange(12), 3))
```

```python
print(np.array_split(np.arange(10), 3))
```

```python
print(np.hsplit(np.arange(12).reshape(3, 4), 2))
```

```python
print(np.vsplit(np.arange(12).reshape(3, 4), 3))
```
