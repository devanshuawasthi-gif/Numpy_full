ARRAY INFORMATION AND DATATYPES

```python
import numpy as np
```

```python
a = np.array([[1,2,3],[4,5,6]], dtype=np.int32)
```

```python
print(a.ndim)
```

```python
print(a.shape)
```

```python
print(a.size)
```

```python
print(a.dtype)
```

```python
print(a.itemsize)
```

```python
x = np.array([1,2,3], dtype=float)
```

```python
y = np.array([1.2,2.5,3.9])
```

```python
print(x.dtype)
```

```python
print(y.astype(int))
```

```python
print(np.array([True, False, True],dtype=bool))
```
