ITERATION

```python
import numpy as np
```

```python
a = np.array([10,20,30])
for item in a:
    print(item)
```

```python
b = np.array([[1,2],[3,4]])
for row in b:
    print(row)
```

```python
for item in np.nditer(b):
    print(item)
```

```python
#enumerate with indices
for index, value in enumerate(a):
    print(index, value)
```
