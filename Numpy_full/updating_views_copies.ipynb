UPDATING , VIEWS, AND COPIES

```python
import numpy as np
```

```python
a = np.array([1,2,3,4])
```

```python
a[0] = 100
```

```python
print(a)
```

# a slice may be a view of original data

```python
x = np.array([10,20,30,40])
```

```python
view = x[1:3]
```

```python
view[0] = 999
```

```python
print(x)
```

#copy() makes an independent copy

```python
x = np.array([10, 20, 30, 40])
```

```python
copied = x[1:3].copy()
```

```python
copied[0] = 999
```

```python
print(x)
```
