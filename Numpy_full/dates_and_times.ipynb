# DATES AND TIMES

```python
import numpy as np
```

```python
dates = np.array(['2026-01-01','2026-01-15','2026-02-01'], dtype='datetime64[D]')
```

```python
print(dates)
```

```python
print(dates[1] - dates[0]) #days between dates
```

```python
print(dates + np.timedelta64(7, 'D')) #add seven days
```
