# FILE INPUT AND OUTPUT

```python
import numpy as np
```

```python
a = np.array([[1,2,3],[4,5,6]])
```

```python
#Numpy nbinary file:  preserves array shape and dtype
np.save('array_data.npy', a)
loaded = np.load('array_data.npy')
print(loaded)
```

```python
#csv/ text file
np.savetxt('array_data.csv', a, delimiter=',',fmt='%d')
loaded_csv = np.loadtxt('array_data.csv',delimiter=',',dtype=int)
print(loaded_csv)
```
