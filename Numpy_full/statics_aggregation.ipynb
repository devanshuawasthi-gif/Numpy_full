# STATISTICS AND AGGREGATIONS

```python
import numpy as np
```

```python
a= np.array([10,20,30,40,50])
```

```python
print(np.sum(a))
```

```python
print(np.mean(a))
```

```python
print(np.median(a))
```

```python
print(np.min(a))
```

```python
print(np.max(a))
```

```python
print(np.ptp(a)) # maximum - minimum
```

```python
print(np.std(a)) #population standarad deviation
```

```python
print(np.var(a))
```

```python
print(np.percentile(a, 25))
```

```python
print(np.quantile(a, 0.75))
```

```python
print(np.prod(a))
```

```python
print(np.cummsum(a))
```

```python
print(np.average(a, weights = [1, 1, 1, 1, 2]))
```

```python
b = np.array([[1,2,3],[4,5,6]])
```

```python
print(np.sum(b, axis = 0)) # per column
```

```python
print(np.sum(b, axis=1)) # per row
```

```python
print(np.mean(b, axis=0))
```

```python
print(np.max(b, axis=1))
```
