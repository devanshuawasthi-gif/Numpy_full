# LINEAR ALGEBRAAAA

```python
import numpy as np
```

```python
a = np.array([[1,2], [3,4]])
```

```python
b = np.array([[5,6],[7,8]])
```

```python
print(a.T) #transpose
```

```python
print(a @ b) # matrix mulltiplication
```

```python
print(np.matmul(a, b))
```

```python
print(np.dot(a, b))
```

```python
print(np.linalg.det(a)) # determinant
```

```python
print(np.linalg.inv(a)) #inverse , if it exist
```

```python
print(np.linalg.matrix_rank(a))
```

```python
print(np.linalg.norm(a)) # norm
```

```python
print(np.linalg.eig(a)) #eignvalue/eigenvectors
```

```python
print(np.linalg.solve(a, np.array([5, 11])))  #solve a @ x = rhs
```
