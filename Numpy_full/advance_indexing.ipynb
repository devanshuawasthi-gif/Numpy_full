# ADVANCE INDEXING

```python
import numpy as np
```

```python
a = np.array([10,20,30,40,50])
```

```python
indices = [0, 2, 4]
```

```python
print(a[indices]) # fancy integer indexing
```

```python
b = np.array([[10, 20],[30,40],[50,60]])
```

```python
print(b[[0,2]]) # first and third rows
```

```python
print(b[[0,1,2], [1,0,1]]) #paired row/column indices
```

```python
#chnage selected values
a[[0, 2]] = [100, 300]
```

```python
print(a)
```

```python
#bollean indexing
print(a[a > 100])
```
