# RANDOM NUMBER GENERATION

```python
import numpy as np
```

```python
rng = np.random.default_rng(42)   #recommended API
```

```python
print(rng.random()) #float[0, 1]
```

```python
print(rng.random((2, 3)))
```

```python
print(rng.integers(1, 7, size=10)) # interger 1 through 6
```

```python
print(rng.normal(0, 1, size=5))
```

```python
print(rng.choice([10, 20, 30], size=5))
```

```python
print(rng.permutation([1, 2, 3, 4]))
```

```python
items = np.array([1,2,3,4,5])
```

```python
rng.shuffle(items)
```

```python
print(items)
```
