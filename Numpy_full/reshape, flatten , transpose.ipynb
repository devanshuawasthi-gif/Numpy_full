# RESHAPE , FLATTEN , TRANSPOSE

```python
import numpy as np
```

```python
a = np.arange(1, 13)
```

```python
print(a.reshape(3, 4))
```

```python
print(a.reshape(2, 2, 3))
```

```python
print(a.reshape(2, -1))
```

```python
print(a.reshape(3,4).flatten())
```

```python
print(a.reshape(3, 4).ravel())
```

```python
print(np.expand_dims(a, axis=0).shape)
```

```python
print(np.expand_dims(a, axis=1).shape)
```

```python
print(np.squeeze(np.array([[[5]]])).shape)
```
