# set operation

```python
import numpy as np
```

```python
a = np.array([1,2,2,3,4])
```

```python
b = np.array([3,4,4,5,6])
```

```python
print(np.unique(a))
```

```python
print(np.union1d(a, b))
```

```python
print(np.intersect1d(a, b))
```

```python
print(np.setdiff1d(a, b)) #value in a but not b
```

```python
print(np.setxor1d(a, b)) # vallue in either, not both
```
