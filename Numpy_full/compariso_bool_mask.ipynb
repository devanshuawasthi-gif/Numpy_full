COMPARISON AND BOOLEAN MARSKS
#use comparison wise elements
#put each comparison inside parantheses

```python
import numpy as np
```

```python
a = np.array([10,20,30,40,50,60,70,80,90])
```

```python
print(a > 25)
```

```python
print(a == 30)
```

```python
print(a[a > 25]) # filter values
```

```python
print(a[(a >= 20) & ( a <= 40)]) # and
```

```python
print(a[(a < 20) | (a > 40)]) # or
```

```python
print(np.any(a > 45))
```

```python
print(np.all(a > 0))
```

```python
print(np.where(a > 25)) #indices
```

```python
print(np.where(a > 25, a, 0)) # keep true values else zeros
```

```python
print(np.count_nonzero(a > 25))
```
