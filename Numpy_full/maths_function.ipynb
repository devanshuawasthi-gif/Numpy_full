# MATHEMATICAL FUNCTIONS

```python
import numpy as np
```

```python
a = np.array([1,4,9,6])
```

```python
print(np.sqrt(a))
```

```python
print(np.square(a))
```

```python
print(np.power(a, 2))
```

```python
print(np.abs([-2, -5, 3]))
```

```python
print(np.exp([0, 1]))
```

```python
print(np.log([1, np.e]))
```

```python
print(np.log10([1, 10, 100]))
```

```python
print(np.round([1.234, 5.678], 2))
```

```python
print(np.floor([1.2, 2.9]))
```

```python
print(np.ceil([1.2, 2.1]))
```

```python
print(np.clip([-5, 3,20], 0, 10))
```

```python
print(np.gcd([12, 20], [8,15]))
```

```python
print(np.lcm([4, 6], [10, 8]))
```

```python
print(np.deg2rad([0, 90, 180]))
```

```python
print(np.rad2deg([0, np.pi / 2]))
```

```python
print(np.sin(np.deg2rad([0, 30, 90])))
```
